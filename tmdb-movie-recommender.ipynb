```python
movies = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_movies.csv')
credits = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_credits.csv')
```

```python
credits.rename(columns = {'movie_id':'id'},inplace = True)
```

```python
credits.head(1)
```

```python
data = pd.merge(movies,credits[[c for c in credits.columns if c not in movies.columns or c == 'id']],on = 'id' )
```

```python
data = data[['id','genres','overview','spoken_languages','title','cast','crew','keywords']]
```

```python
data.head(2)
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace = True)
```

```python
data.duplicated().sum()
```

```python
data.head(2)
```

```python
import ast
def convert(text):
    l = []
    for i in ast.literal_eval(text):
        l.append(i['name'])
    return l
```

```python
data['genres'] = data['genres'].apply(convert)
data.head(2)
```

```python
data['keywords'] = data['keywords'].apply(convert)
data.head(2)
```

```python
data['spoken_languages'] = data['spoken_languages'].apply(convert)
data.head(2)
```

```python
def convert_cast(text):
    l = []
    count = 0
    for i in ast.literal_eval(text):
        if count<3:
            l.append(i['name'])
        count+=1
    return l
```

```python
data['cast'] = data['cast'].apply(convert_cast)
data.head(2)
```

```python
def director(text):
    l = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l
```

```python
data['crew'] = data['crew'].apply(director)
data.head(2)
```

```python
data['overview'] = data['overview'].apply(lambda x:x.split())
data.head(2)
```

```python
def remove_space(text):
    l = []
    for i in text:
        l.append(i.replace(" ",""))
    return l
```

```python
data['cast'] = data['cast'].apply(remove_space)
data['crew'] = data['crew'].apply(remove_space)
data['genres'] = data['genres'].apply(remove_space)
data['keywords'] = data['keywords'].apply(remove_space)
```

```python
data.head(2)
```

```python
data['tags'] = data['overview']+data['genres']+data['spoken_languages']+data['cast']+data['crew']+data['keywords']
```

```python
data.head(2)
```

```python
data = data[['id','title','tags']] 
```

```python
data.head(2)
```

```python
data['tags'] = data['tags'].apply(lambda x:" ".join(x))
data.head()
```

```python
data['tags'] = data['tags'].apply(lambda x:x.lower())
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stems(text):
    l = []
    for i in text.split():
        l.append(ps.stem(i))
    return " ".join(l)
```

```python
data['tags'] = data['tags'].apply(stems)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000,stop_words = 'english')
```

```python
vector = cv.fit_transform(data['tags']).toarray()
vector = np.array(vector, dtype=np.float32)
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarly = cosine_similarity(vector)
similarly
```

```python
similarly.shape
```

```python
def recommend(movie):
    index = data[data['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarly[index])),reverse = True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(data.iloc[i[0]].title)
```

```python
recommend('Spider-Man')
```
